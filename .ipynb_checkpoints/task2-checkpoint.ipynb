```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "tabularisai/multilingual-sentiment-analysis"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

def predict_sentiment(texts):
    inputs = tokenizer(texts, return_tensors="pt", truncation=True, padding=True, max_length=512)
    with torch.no_grad():
        outputs = model(**inputs)
    probabilities = torch.nn.functional.softmax(outputs.logits, dim=-1)
    sentiment_map = {0: "Very Negative", 1: "Negative", 2: "Neutral", 3: "Positive", 4: "Very Positive"}
    return [sentiment_map[p] for p in torch.argmax(probabilities, dim=-1).tolist()]

texts = [
    "Мені не просто сподобався Аркейн, а я був у захваті!", "Обслуговування було жахливим, але в цьому полягає фішка даного ресторану, тому все було чудово, хоча й доволі грубо.",
    "Гра Престолів була чудовою, але останній сезон все зіпсував",
    "Ця гра непогана, якщо у вас проблеми з інтелектом", "Ця гра на один раз, я не розумію, як вона стала грою року",
    "Ця гра нормальна, якщо не звертати увагу на графіку, сюжет, персонажів, музику, озвучку, геймплей. Простіше кажучи, якщо просто її не запускати",
    "Багато хто хейтить трейер Відьмака4, через вигляд Цирі, але я находжу його доволі вдалим",
    "Обслуговування було жахливим.",
    "Книга була посередньою, хто ж знав що книги потрібно читати, це так довго...."
]

for text, sentiment in zip(texts, predict_sentiment(texts)):
    print(f"Комментар: {text}\n Відгук: {sentiment}\n")
```
